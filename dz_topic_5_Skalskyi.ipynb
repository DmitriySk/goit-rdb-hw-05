## Завдання 2. CTE-chain pipeline з ≥5 stages

Pipeline зробив з 8 stages (0-7), кожен робить одну річ:

- stage 0 `reference_date` - дата відліку, MAX(order_purchase_timestamp) + 1 день
- stage 1 `stage1_orders_total` - сума кожного замовлення (price + freight_value)
- stage 2 `stage2_rfm` - recency_days, frequency, monetary і дата першого замовлення
- stage 3 `stage3_cohort` - cohort_quarter
- stage 4 `stage4_rolling_aov` - rolling AVG за 30 днів через window function
- stage 5 `stage5_cancel_rate` - частка скасованих замовлень
- stage 6 `stage6_category_revenue` - виручка клієнта по категоріях
- stage 7 `stage7_preferred_category` - топ-1 категорія через ROW_NUMBER()

recency_days рахую від reference_date, а не від CURRENT_DATE, бо дані старі (2016-2018) і з CURRENT_DATE результат мінявся б кожен день. В кінці роблю LEFT JOIN від olist_customers, щоб не загубити клієнтів.

```python
show("""
    WITH
    reference_date AS (
        -- Stage 0: stable reference date for historical dataset
        SELECT (MAX(order_purchase_timestamp)::DATE + 1) AS as_of_date
        FROM olist_orders
    ),
    stage1_orders_total AS (
        -- Stage 1: orders + order_total + status
        SELECT
            o.customer_id,
            o.order_id,
            o.order_status,
            o.order_purchase_timestamp AS order_ts,
            o.order_purchase_timestamp::DATE AS order_date,
            COALESCE(SUM(
                COALESCE(oi.price, 0) + COALESCE(oi.freight_value, 0)
            ), 0)::NUMERIC(12, 2) AS order_total
        FROM olist_orders AS o
        LEFT JOIN olist_order_items AS oi
            ON oi.order_id = o.order_id
        GROUP BY
            o.customer_id,
            o.order_id,
            o.order_status,
            o.order_purchase_timestamp
    ),
    stage2_rfm AS (
        -- Stage 2: RFM features per customer
        SELECT
            s.customer_id,
            (rd.as_of_date - MAX(s.order_date))::INTEGER AS recency_days,
            COUNT(*) AS frequency,
            SUM(s.order_total)::NUMERIC(12, 2) AS monetary,
            MIN(s.order_date) AS first_order_date
        FROM stage1_orders_total AS s
        CROSS JOIN reference_date AS rd
        GROUP BY s.customer_id, rd.as_of_date
    ),
    stage3_cohort AS (
        -- Stage 3: cohort quarter from first order date
        SELECT
            r.*,
            DATE_TRUNC('quarter', r.first_order_date)::DATE AS cohort_quarter
        FROM stage2_rfm AS r
    ),
    stage4_rolling_aov AS (
        -- Stage 4: rolling 30-day average order value per customer
        SELECT
            s.customer_id,
            s.order_id,
            s.order_ts,
            s.order_total,
            AVG(s.order_total) OVER (
                PARTITION BY s.customer_id
                ORDER BY s.order_ts
                RANGE BETWEEN INTERVAL '30 days' PRECEDING AND CURRENT ROW
            )::NUMERIC(12, 2) AS avg_order_value_30d,
            ROW_NUMBER() OVER (
                PARTITION BY s.customer_id
                ORDER BY s.order_ts DESC, s.order_id DESC
            ) AS rn_recent
        FROM stage1_orders_total AS s
    ),
    stage5_cancel_rate AS (
        -- Stage 5: cancellation rate per customer
        SELECT
            customer_id,
            AVG(
                CASE WHEN order_status = 'canceled' THEN 1.0 ELSE 0.0 END
            )::NUMERIC(6, 4) AS cancellation_rate
        FROM stage1_orders_total
        GROUP BY customer_id
    ),
    stage6_category_revenue AS (
        -- Stage 6: revenue by customer and product category
        SELECT
            o.customer_id,
            COALESCE(p.product_category_name, '(unknown)') AS product_category_name,
            SUM(
                COALESCE(oi.price, 0) + COALESCE(oi.freight_value, 0)
            )::NUMERIC(12, 2) AS category_revenue
        FROM olist_orders AS o
        JOIN olist_order_items AS oi
            ON oi.order_id = o.order_id
        LEFT JOIN olist_products AS p
            ON p.product_id = oi.product_id
        GROUP BY o.customer_id, COALESCE(p.product_category_name, '(unknown)')
    ),
    stage7_preferred_category AS (
        -- Stage 7: preferred category per customer via top-1 ranking
        SELECT
            customer_id,
            product_category_name,
            category_revenue
        FROM (
            SELECT
                cr.*,
                ROW_NUMBER() OVER (
                    PARTITION BY cr.customer_id
                    ORDER BY cr.category_revenue DESC, cr.product_category_name
                ) AS rn
            FROM stage6_category_revenue AS cr
        ) AS ranked
        WHERE rn = 1
    )
    SELECT
        c.customer_id,
        c.customer_state,
        COALESCE(coh.recency_days, 9999) AS recency_days,
        COALESCE(coh.frequency, 0) AS frequency,
        COALESCE(coh.monetary, 0)::NUMERIC(12, 2) AS monetary,
        coh.first_order_date,
        coh.cohort_quarter,
        COALESCE(aov.avg_order_value_30d, 0)::NUMERIC(12, 2) AS avg_order_value_30d_recent,
        COALESCE(cancel.cancellation_rate, 0)::NUMERIC(6, 4) AS cancellation_rate,
        COALESCE(pc.product_category_name, '(none)') AS preferred_category,
        COALESCE(pc.category_revenue, 0)::NUMERIC(12, 2) AS preferred_category_revenue
    FROM olist_customers AS c
    LEFT JOIN stage3_cohort AS coh
        ON coh.customer_id = c.customer_id
    LEFT JOIN stage4_rolling_aov AS aov
        ON aov.customer_id = c.customer_id
       AND aov.rn_recent = 1
    LEFT JOIN stage5_cancel_rate AS cancel
        ON cancel.customer_id = c.customer_id
    LEFT JOIN stage7_preferred_category AS pc
        ON pc.customer_id = c.customer_id
    ORDER BY monetary DESC NULLS LAST, c.customer_id
    LIMIT 20;
""")
```

**Висновок:**

Всі потрібні ознаки є: recency_days, frequency, monetary, cohort_quarter, avg_order_value_30d_recent, cancellation_rate, preferred_category. Найбільший monetary 13664.08, категорія telefonia_fixa.

reference_date виходить 2018-10-18 (останнє замовлення 17.10.2018), тому результат не залежить від того, коли запускати ноутбук.

Помітив, що в топ-20 у всіх frequency = 1. Як я зрозумів, в Olist customer_id створюється окремо на кожне замовлення, а одну людину позначає customer_unique_id. Через це avg_order_value_30d_recent тут співпадає з monetary, а cancellation_rate буває тільки 0 або 1. В завданні ключ customer_id, тому залишив так.

Ще в рядку 7 cancellation_rate = 1, тобто замовлення скасоване, але воно все одно рахується в monetary.

## Завдання 3. Recursive CTE для referral chain

В Olist немає таблиці рефералів, тому створюю свою `hw5_referrals`. Беру перших 10 клієнтів по customer_id і роблю з них ланцюжок: кожен запрошує наступного. PK на referred_customer_id - щоб у клієнта був тільки один referrer, CHECK - щоб не можна було запросити самого себе.

```python
run("""
    DROP TABLE IF EXISTS hw5_referrals CASCADE;

    CREATE TABLE hw5_referrals (
        referrer_customer_id TEXT NOT NULL REFERENCES olist_customers(customer_id),
        referred_customer_id TEXT NOT NULL REFERENCES olist_customers(customer_id),
        referred_at          TIMESTAMPTZ NOT NULL DEFAULT NOW(),
        PRIMARY KEY (referred_customer_id),
        CHECK (referrer_customer_id <> referred_customer_id)
    );

    WITH ordered_customers AS (
        SELECT
            customer_id,
            ROW_NUMBER() OVER (ORDER BY customer_id) AS rn
        FROM olist_customers
        ORDER BY customer_id
        LIMIT 10
    )
    INSERT INTO hw5_referrals (referrer_customer_id, referred_customer_id)
    SELECT
        prev.customer_id AS referrer_customer_id,
        nxt.customer_id AS referred_customer_id
    FROM ordered_customers AS prev
    JOIN ordered_customers AS nxt
        ON nxt.rn = prev.rn + 1
    ON CONFLICT (referred_customer_id) DO NOTHING;
""")
```

```python
show("""
    SELECT *
    FROM hw5_referrals
    ORDER BY referrer_customer_id, referred_customer_id;
""")
```

Рекурсивний запит:

- anchor - клієнти, яких ніхто не запрошував (через NOT EXISTS, бо NOT IN може зламатись якщо є NULL)
- recursive part - йдемо від referrer до тих, кого він запросив, depth + 1 і дописуємо path
- захист від циклів - `NOT r.referred_customer_id = ANY(rd.path_ids)`, щоб не зайти в того самого клієнта ще раз
- depth guard - `rd.depth < 50`
- рекурсія закінчується, коли recursive part нічого нового не повертає

```python
show("""
    WITH RECURSIVE referral_depth AS (
        -- ANCHOR: root customers, які не були referred іншими customers
        SELECT
            c.customer_id,
            1 AS depth,
            ARRAY[c.customer_id]::TEXT[] AS path_ids,
            c.customer_id::TEXT AS path_text
        FROM olist_customers AS c
        WHERE NOT EXISTS (
            SELECT 1
            FROM hw5_referrals AS r
            WHERE r.referred_customer_id = c.customer_id
        )

        UNION ALL

        -- RECURSIVE: перейти від referrer до його referred customers
        SELECT
            r.referred_customer_id AS customer_id,
            rd.depth + 1 AS depth,
            rd.path_ids || r.referred_customer_id AS path_ids,
            rd.path_text || ' -> ' || r.referred_customer_id AS path_text
        FROM hw5_referrals AS r
        JOIN referral_depth AS rd
            ON rd.customer_id = r.referrer_customer_id
        WHERE NOT r.referred_customer_id = ANY(rd.path_ids)
          AND rd.depth < 50
    )
    SELECT
        depth,
        COUNT(*) AS n_customers,
        MAX(path_text) AS sample_path
    FROM referral_depth
    GROUP BY depth
    ORDER BY depth;
""")
```

**Висновок:** на depth = 1 вийшло 99432 клієнти, це всі клієнти (99441) мінус 9 запрошених. Далі по одному клієнту на кожному рівні від 2 до 10, на depth = 10 видно весь ланцюжок з 10 customer_id. Останній в ланцюжку нікого не запрошував, тому рекурсія сама зупинилась, до ліміту 50 не дійшло.

## Завдання 4. Materialized View з CONCURRENT REFRESH

В MV кладу весь pipeline із завдання 2 (ті самі stages і колонки), тільки без ORDER BY і LIMIT, щоб там були всі клієнти.

```python
run("""
    DROP MATERIALIZED VIEW IF EXISTS mv_olist_customer_features CASCADE;

    CREATE MATERIALIZED VIEW mv_olist_customer_features AS
    WITH
    reference_date AS (
        -- Stage 0: stable reference date for historical dataset
        SELECT (MAX(order_purchase_timestamp)::DATE + 1) AS as_of_date
        FROM olist_orders
    ),
    stage1_orders_total AS (
        -- Stage 1: orders + order_total + status
        SELECT
            o.customer_id,
            o.order_id,
            o.order_status,
            o.order_purchase_timestamp AS order_ts,
            o.order_purchase_timestamp::DATE AS order_date,
            COALESCE(SUM(
                COALESCE(oi.price, 0) + COALESCE(oi.freight_value, 0)
            ), 0)::NUMERIC(12, 2) AS order_total
        FROM olist_orders AS o
        LEFT JOIN olist_order_items AS oi
            ON oi.order_id = o.order_id
        GROUP BY
            o.customer_id,
            o.order_id,
            o.order_status,
            o.order_purchase_timestamp
    ),
    stage2_rfm AS (
        -- Stage 2: RFM features per customer
        SELECT
            s.customer_id,
            (rd.as_of_date - MAX(s.order_date))::INTEGER AS recency_days,
            COUNT(*) AS frequency,
            SUM(s.order_total)::NUMERIC(12, 2) AS monetary,
            MIN(s.order_date) AS first_order_date
        FROM stage1_orders_total AS s
        CROSS JOIN reference_date AS rd
        GROUP BY s.customer_id, rd.as_of_date
    ),
    stage3_cohort AS (
        -- Stage 3: cohort quarter from first order date
        SELECT
            r.*,
            DATE_TRUNC('quarter', r.first_order_date)::DATE AS cohort_quarter
        FROM stage2_rfm AS r
    ),
    stage4_rolling_aov AS (
        -- Stage 4: rolling 30-day average order value per customer
        SELECT
            s.customer_id,
            s.order_id,
            s.order_ts,
            s.order_total,
            AVG(s.order_total) OVER (
                PARTITION BY s.customer_id
                ORDER BY s.order_ts
                RANGE BETWEEN INTERVAL '30 days' PRECEDING AND CURRENT ROW
            )::NUMERIC(12, 2) AS avg_order_value_30d,
            ROW_NUMBER() OVER (
                PARTITION BY s.customer_id
                ORDER BY s.order_ts DESC, s.order_id DESC
            ) AS rn_recent
        FROM stage1_orders_total AS s
    ),
    stage5_cancel_rate AS (
        -- Stage 5: cancellation rate per customer
        SELECT
            customer_id,
            AVG(
                CASE WHEN order_status = 'canceled' THEN 1.0 ELSE 0.0 END
            )::NUMERIC(6, 4) AS cancellation_rate
        FROM stage1_orders_total
        GROUP BY customer_id
    ),
    stage6_category_revenue AS (
        -- Stage 6: revenue by customer and product category
        SELECT
            o.customer_id,
            COALESCE(p.product_category_name, '(unknown)') AS product_category_name,
            SUM(
                COALESCE(oi.price, 0) + COALESCE(oi.freight_value, 0)
            )::NUMERIC(12, 2) AS category_revenue
        FROM olist_orders AS o
        JOIN olist_order_items AS oi
            ON oi.order_id = o.order_id
        LEFT JOIN olist_products AS p
            ON p.product_id = oi.product_id
        GROUP BY o.customer_id, COALESCE(p.product_category_name, '(unknown)')
    ),
    stage7_preferred_category AS (
        -- Stage 7: preferred category per customer via top-1 ranking
        SELECT
            customer_id,
            product_category_name,
            category_revenue
        FROM (
            SELECT
                cr.*,
                ROW_NUMBER() OVER (
                    PARTITION BY cr.customer_id
                    ORDER BY cr.category_revenue DESC, cr.product_category_name
                ) AS rn
            FROM stage6_category_revenue AS cr
        ) AS ranked
        WHERE rn = 1
    )
    SELECT
        c.customer_id,
        c.customer_state,
        COALESCE(coh.recency_days, 9999) AS recency_days,
        COALESCE(coh.frequency, 0) AS frequency,
        COALESCE(coh.monetary, 0)::NUMERIC(12, 2) AS monetary,
        coh.first_order_date,
        coh.cohort_quarter,
        COALESCE(aov.avg_order_value_30d, 0)::NUMERIC(12, 2) AS avg_order_value_30d_recent,
        COALESCE(cancel.cancellation_rate, 0)::NUMERIC(6, 4) AS cancellation_rate,
        COALESCE(pc.product_category_name, '(none)') AS preferred_category,
        COALESCE(pc.category_revenue, 0)::NUMERIC(12, 2) AS preferred_category_revenue
    FROM olist_customers AS c
    LEFT JOIN stage3_cohort AS coh
        ON coh.customer_id = c.customer_id
    LEFT JOIN stage4_rolling_aov AS aov
        ON aov.customer_id = c.customer_id
       AND aov.rn_recent = 1
    LEFT JOIN stage5_cancel_rate AS cancel
        ON cancel.customer_id = c.customer_id
    LEFT JOIN stage7_preferred_category AS pc
        ON pc.customer_id = c.customer_id;

    CREATE UNIQUE INDEX idx_mv_olist_customer_features_customer_id
        ON mv_olist_customer_features (customer_id);
""")
```

## Завдання 5. EXPLAIN ANALYZE + Reflection

### 1. EXPLAIN на повному CTE-pipeline

```python
explain("""
    EXPLAIN (ANALYZE, BUFFERS)
    WITH
    reference_date AS (
        SELECT (MAX(order_purchase_timestamp)::DATE + 1) AS as_of_date
        FROM olist_orders
    ),
    stage1_orders_total AS (
        SELECT
            o.customer_id,
            o.order_id,
            o.order_status,
            o.order_purchase_timestamp AS order_ts,
            o.order_purchase_timestamp::DATE AS order_date,
            COALESCE(SUM(
                COALESCE(oi.price, 0) + COALESCE(oi.freight_value, 0)
            ), 0)::NUMERIC(12, 2) AS order_total
        FROM olist_orders AS o
        LEFT JOIN olist_order_items AS oi
            ON oi.order_id = o.order_id
        GROUP BY
            o.customer_id,
            o.order_id,
            o.order_status,
            o.order_purchase_timestamp
    ),
    stage2_rfm AS (
        SELECT
            s.customer_id,
            (rd.as_of_date - MAX(s.order_date))::INTEGER AS recency_days,
            COUNT(*) AS frequency,
            SUM(s.order_total)::NUMERIC(12, 2) AS monetary,
            MIN(s.order_date) AS first_order_date
        FROM stage1_orders_total AS s
        CROSS JOIN reference_date AS rd
        GROUP BY s.customer_id, rd.as_of_date
    )
    SELECT *
    FROM stage2_rfm
    ORDER BY monetary DESC NULLS LAST
    LIMIT 10;
""")
```

### 2. EXPLAIN на lookup із MV

```python
explain("""
    EXPLAIN (ANALYZE, BUFFERS)
    WITH target_customer AS (
        SELECT customer_id
        FROM olist_customers
        ORDER BY customer_id
        LIMIT 1
    )
    SELECT mv.*
    FROM mv_olist_customer_features AS mv
    JOIN target_customer AS t
        ON t.customer_id = mv.customer_id;
""")
```

### 3. Додатково: top-N з MV + індекс на monetary

```python
run("""
    CREATE INDEX IF NOT EXISTS idx_mv_olist_customer_features_monetary
        ON mv_olist_customer_features (monetary DESC);
""")

explain("""
    EXPLAIN (ANALYZE, BUFFERS)
    SELECT *
    FROM mv_olist_customer_features
    ORDER BY monetary DESC NULLS LAST
    LIMIT 10;
""")
```

### Що видно в планах

1. Pipeline (до RFM) - **327.8 ms**. В плані Seq Scan по olist_orders і olist_order_items, Parallel Hash Right Join, потім Sort по order_id, який не влазить в пам'ять і пише на диск (external sort 6600kB + 5992kB у воркера), GroupAggregate по order_id і ще HashAggregate по customer_id (Batches: 21, Disk Usage: 7400kB). temp read=3356 written=4096, тобто запит реально працював з диском.
2. Lookup з MV - **0.107 ms**. Index Scan по idx_mv_olist_customer_features_customer_id, всього 8 буферів (shared hit=5 read=3).
3. Top-N з MV - **19.9 ms**. Parallel Seq Scan по MV + top-N heapsort.

Виходить, що lookup з MV приблизно в 3000 разів швидший, ніж перерахунок навіть частини pipeline.

В 3-му запиті індекс на monetary не використався. Індекс `(monetary DESC)` за замовчуванням має NULLS FIRST, а в запиті `DESC NULLS LAST`, тому порядок не співпадає і планувальник робить Seq Scan. Треба або створити індекс як `(monetary DESC NULLS LAST)`, або прибрати NULLS LAST з запиту (NULL в monetary все одно нема через COALESCE).

## Reflection

**1. Найдорожчий feature.** По EXPLAIN найдорожчий monetary (RFM), а точніше stage1, де рахується order_total. Там JOIN orders і order_items (Parallel Hash Right Join), потім Sort по order_id з записом на диск і GroupAggregate, це вже ~210 ms з 328. Після цього ще HashAggregate по customer_id, який теж не вліз в пам'ять (21 batch, 7400kB на диск). Тобто основний час йде на Sort і Hash Aggregate. А ще rolling AOV і preferred_category додають window functions зі своїми сортуваннями, тому повний pipeline ще повільніший.

**2. Чи кешувати через MV.** Так, варто. Читання з MV по customer_id це Index Scan за 0.1 ms і 8 буферів, а перерахунок 300+ ms. Дані в Olist історичні і оновлюються пачками, тому я б робив REFRESH CONCURRENTLY раз на добу (вночі після завантаження нових даних) і ще on-demand, якщо щось перезаливали. Щогодини сенсу нема, recency_days все одно в днях.

**3. Коли краще звичайний VIEW.** Коли потрібні завжди актуальні дані, а запитів мало, наприклад для разової аналітики чи перевірки даних. Або коли таблиці маленькі і перерахунок і так швидкий.

**4. Termination в recursive CTE.** Рекурсія зупиняється, коли recursive part повертає 0 рядків. Якщо є цикл A → B → A і захисту нема, то UNION ALL буде ходити по колу, поки запит не впаде по таймауту. У мене два захисти: `NOT ... = ANY(path_ids)` (A вже є в path, тому назад в нього не підемо) і `depth < 50`. Ще можна використати CYCLE в PostgreSQL 14+.

**5. Trade-off.** Як на мене, тут головне recompute-cost vs latency: перерахунок займає сотні мілісекунд і буде рости разом з даними, а для serving треба відповідати швидко. Freshness не так важлива, бо дані історичні, а storage маленький (~99 тис. рядків в MV).
